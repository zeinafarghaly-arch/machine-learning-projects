```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf
from tensorflow.keras import layers, models
```

```python
(X_train, y_train), (X_test, y_test) = tf.keras.datasets.fashion_mnist.load_data()
```

```python
X_train=X_train.astype('float32')/255
X_test=X_test.astype('float32')/255
```

```python
x_train_flatten=X_train.reshape(-1, 28*28)
x_test_flatten=X_test.reshape(-1, 28*28)
```

```python
model_A=models.Sequential()
```

```python
model_A.add(layers.Input(shape=(28*28,)))
model_A.add(layers.Dense(128, activation= 'relu'))
model_A.add(layers.Dense(64, activation= 'relu'))
model_A.add(layers.Dense(10, activation= 'softmax'))
```

```python
model_A.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
early_stop=tf.keras.callbacks.EarlyStopping(
    monitor="val_loss",
    patience=4,
    restore_best_weights=True,
)
```

```python
historyA = model_A.fit(
    x_train_flatten,
    y_train,
    validation_split=0.1,
    epochs=50,
    batch_size=64, 
    callbacks=[early_stop]
)
```

```python
model_B=models.Sequential()
```

```python
model_B.add(layers.Input(shape=(28*28,)))
model_B.add(layers.Dense(256, activation= 'tanh'))
model_B.add(layers.Dense(128, activation= 'tanh'))
model_B.add(layers.Dense(64, activation= 'tanh'))
model_B.add(layers.Dense(10, activation= 'softmax'))
```

```python
model_B.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
historyB = model_B.fit(
    x_train_flatten,
    y_train,
    validation_split=0.1,
    epochs=50,
    batch_size=64, 
    callbacks=[early_stop]
)
```

```python
model_C=models.Sequential()
```

```python
model_C.add(layers.Input(shape=(28*28,)))
model_C.add(layers.Dense(128, activation= 'relu'))
model_C.add(layers.Dense(64, activation= 'relu'))
model_C.add(layers.Dense(10, activation= 'softmax'))
```

```python
model_C.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
historyC = model_C.fit(
    x_train_flatten,
    y_train,
    validation_split=0.1,
    epochs=50,
    batch_size=256, 
    callbacks=[early_stop]
)
```

```python
print("Model A")
print("Best Val Accuracy:", max(historyA.history["val_accuracy"]))
print("Best Val Loss:", min(historyA.history["val_loss"]))
print("Model B")
print("Best Val Accuracy:", max(historyB.history["val_accuracy"]))
print("Best Val Loss:", min(historyB.history["val_loss"]))
print("Model C")
print("Best Val Accuracy:", max(historyC.history["val_accuracy"]))
print("Best Val Loss:", min(historyC.history["val_loss"]))
```

```python
best_model = model_C
history = historyC
```

```python
history = historyC
plt.figure(figsize=(14,5))
plt.subplot(1,2,1)
plt.plot(history.history['accuracy'], marker='o', label='Training Accuracy')
plt.plot(history.history['val_accuracy'], marker='o', label='Validation Accuracy')
plt.title('Training vs Validation Accuracy')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()
plt.grid(True)

plt.subplot(1,2,2)
plt.plot(history.history['loss'], marker='o', label='Training Loss')
plt.plot(history.history['val_loss'], marker='o', label='Validation Loss')
plt.title('Training vs Validation Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()
plt.grid(True)

plt.show()
```

```python
y_pred_probs = model_C.predict(x_test_flatten)
y_pred = np.argmax(y_pred_probs, axis=1)
y_true = y_test
```

```python
from sklearn.metrics import confusion_matrix
class_names = [
    "T-shirt", "Trouser", "Pullover", "Dress", "Coat",
    "Sandal", "Shirt", "Sneaker", "Bag", "Ankle Boot"
]
cm = confusion_matrix(y_true, y_pred)
plt.figure(figsize=(10, 8))
sns.heatmap(
    cm,
    annot=True,
    fmt='d',
    cmap='Blues',
    xticklabels=class_names,
    yticklabels=class_names
)

plt.title("Confusion Matrix - Model C")
plt.xlabel("Predicted Label")
plt.ylabel("True Label")
plt.xticks(rotation=45)
plt.yticks(rotation=0)
plt.show()
```

```python
wrong = np.where(y_pred != y_test)[0]

plt.figure(figsize=(15,6))

for i in range(10):
    plt.subplot(2,5,i+1)
    plt.imshow(X_test[wrong[i]], cmap="gray")
    plt.title(f"True: {class_names[y_test[wrong[i]]]}\nPred: {class_names[y_pred[wrong[i]]]}")
    plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
print("Model A")
model_A.evaluate(x_test_flatten, y_test)

print("Model B")
model_B.evaluate(x_test_flatten, y_test)

print("Model C")
model_C.evaluate(x_test_flatten, y_test)
```
